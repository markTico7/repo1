```python
plt.hist (df['Current'], bins=5,
          alpha = 0.8,
          color = '#126782',
          edgecolor= '#1C1F24')

plt.title('Histogram of Current values',
          font = 'Calibri',
          fontsize = 12,

          )

plt.xlabel('Current (mA)')
plt.ylabel('Frequency')

plt.show()
```

```python
# normal distribution curve
mean = df['Current'].mean()
std = df['Current'].std(ddof=1)

# x-axis
x_min = df['Current'].min()
x_max = df['Current'].max()

x = np.linspace(x_min, x_max, 100)
p = stats.norm.pdf(x, mean, std)

plt.plot(x, p,
         color = '#126782',
         linewidth = 2
         )

plt.title('Normal Distribution of Current Values',
          font = 'Calibri',
          fontsize= 12)

plt.xlabel('Current (mA)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist (df['Current'], bins=5,
          density =True,
          alpha = 0.8,
          color = '#126782',
          edgecolor= '#1C1F24')

plt.title('Normal Distribution of Current Values',
          font = 'Calibri',
          fontsize= 12)
plt.xlabel('Current (mA)')
plt.ylabel('Frequency')



p = stats.norm.pdf(x, mean, std)
plt.plot(x, p,
         color = "#FF0000",
         linewidth = 2
         )

plt.show()
```

```python
# standard normal distribution N(0,1)
# x-axis

x_max = df['Z-score'].max()
x_min = df['Z-score'].min()
x = np.linspace(x_min, x_max, 100)

p = stats.norm.pdf(x, df['Z-score'].mean(), df['Z-score'].std())

plt.plot(x, p,
         color = '#126782',
         linewidth = 2
         )

plt.title('Normal Distribution of Z-score Values',
          font = 'Calibri',
          fontsize= 12)

plt.xlabel('Z-score')
plt.ylabel('Probability Density')
plt.show()
```
